```python
import rclpy as rp
rp.init()
```

```python
Heart_node = rp.create_node('Heart_node')
```

```python
from geometry_msgs.msg import Twist
```

```python
pub = Heart_node.create_publisher(Twist, 'turtle1/cmd_vel', 10)
```

```python
cnt = 0
timer_period = 1
def timer_callback():
    global cnt
    cnt += 1

    msg = Twist()

    if cnt == 1 or (cnt == 4):
        msg.angular.z = 1.570796
        msg.linear.x = 0.0
        print(cnt)
        pub.publish(msg)
    elif (cnt == 2) or (cnt == 6):
        msg.angular.z = 3.92699
        msg.linear.x = 5.0
        print(cnt)
        pub.publish(msg)
    elif (cnt == 3) or (cnt == 5):
        msg.angular.z = 0.0
        msg.linear.x = 3.0738
        print(cnt)
        pub.publish(msg)
    else:
        raise Exception("end")
```

```python
timer = Heart_node.create_timer(timer_period, timer_callback)
```

```python
rp.spin(Heart_node)
```
